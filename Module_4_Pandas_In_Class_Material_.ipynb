### Understanding Pandas Index Objects

In pandas, an `Index` object is like a special list of labels for your data. Think of it as the 'name tags' for rows in a Series or for both rows and columns in a DataFrame. These labels are crucial for easily finding and working with specific parts of your data.

**Key Roles of Index Objects:**
1.  **Labels for Axes**: They provide meaningful names (like 'red', 'blue') instead of just numbers (like 0, 1, 2) for your rows and columns. This makes your data much easier to understand and manipulate.
2.  **Metadata Storage**: They can hold extra information, such as the data type of the labels themselves (e.g., `dtype='object'` for strings).

**Important Characteristics:**
*   **Immutability**: Once an `Index` is created, you cannot change its individual elements directly. This ensures that your data labels remain consistent and don't accidentally get altered, which is important when sharing or referencing the index across different parts of your data.
*   **Useful Methods**: `Index` objects come with handy functions. For instance, `idxmin()` tells you the label of the row with the smallest value, and `idxmax()` tells you the label of the row with the largest value.

Let's see this in action with a simple `Series`:

```python
import pandas as pd

# Create a Series with custom index labels
ser = pd.Series([5, 0, 3, 8, 4], index=['red', 'blue', 'yellow', 'white', 'green'])
print("Series 'ser':\n", ser)

# Accessing the Index object of the Series
series_index = ser.index
print("\nIndex object of 'ser' (ser.index):\n", series_index)
print("Type of series_index:", type(series_index))

# Demonstrate immutability (this line would raise a TypeError if uncommented)
# ser.index[0] = 'crimson' # Uncommenting this line will cause an error!
print("\nIndex objects are IMMUTABLE (cannot be modified in-place once created).")

# Use idxmin() and idxmax() to find labels of min/max values
print("\nLabel of minimum value (ser.idxmin()):", ser.idxmin())
print("Label of maximum value (ser.idxmax()):", ser.idxmax())
```

### Duplicate Labels in Series and DataFrames

While an `Index` usually contains unique labels, pandas allows for duplicate labels. This can be useful in certain data arrangements, but it changes how data selection works.

When you select data using a label that appears multiple times in the index, pandas will return all entries corresponding to that label.

`is_unique` is a useful property to check if all labels in an index are unique.

```python
import pandas as pd
import numpy as np

# Create a Series with duplicate labels
serd = pd.Series(range(6),
                 index=['white', 'white', 'blue', 'green', 'green', 'yellow'])
print("Series 'serd' with duplicate labels:\n", serd)

# Select elements with a duplicate label
duplicate_selection = serd['white']
print("\nSelecting 'white' from 'serd':\n", duplicate_selection)

# Check if index has unique labels
print("\nIs 'serd' index unique? (serd.index.is_unique):", serd.index.is_unique)

# For comparison, check a DataFrame with a unique default index
data = {'col1': [1,2,3], 'col2': [4,5,6]}
frame_unique = pd.DataFrame(data)
print("\nIs 'frame_unique' index unique? (frame_unique.index.is_unique):", frame_unique.index.is_unique)
```

### Reindexing: Changing the Index of a Series or DataFrame

Although `Index` objects themselves are immutable (meaning you can't change individual labels after creation), pandas provides a powerful way to change the index of a Series or DataFrame: **reindexing**.

The `reindex()` method allows you to create a *new* object with a different index. You can:
*   Change the order of existing labels.
*   Add new labels (which will result in `NaN` for their values by default).
*   Remove existing labels.

#### Handling Missing Values During Reindexing
When you add new labels that weren't present in the original data, pandas fills them with `NaN` (Not a Number). You can use `method` arguments to specify how these missing values should be handled, such as `ffill` (forward fill) or `bfill` (backward fill) to propagate existing values.

```python
import pandas as pd
import numpy as np

ser3 = pd.Series([1,5,6,3],index=[0,3,5,6])
print("Original Series 'ser3':\n", ser3)

# Reindex 'ser3' to a new range of indices (0 to 5)
# New indices 1, 2, 4 are added and filled with NaN by default
print("\nReindexed 'ser3' (default NaN fill):\n", ser3.reindex(range(6)))

# Reindex using 'bfill' (backward fill): values are filled from the next valid observation
print("\nReindexed 'ser3' with 'bfill':\n", ser3.reindex(range(6), method='bfill'))

# Reindex using 'ffill' (forward fill): values are filled from the previous valid observation
print("\nReindexed 'ser3' with 'ffill':\n", ser3.reindex(range(6), method='ffill'))

# Reindexing a DataFrame
data = {
    'color':['blue','green','yellow','orange','white'],
    'object':['ball','pen','pencil','paper','sharpner'],
    'price':[1.2,1.0,0.6,0.9,1.7]
}
frame = pd.DataFrame(data,
                      index = ['one','two','three','four','five'])
print("\nOriginal DataFrame 'frame':\n", frame)

# Reindex DataFrame rows and add a new column 'new'
# The 'method='ffill'' was removed from the row reindex to prevent the ValueError.
# It will now introduce NaN values for the new row 'six'.
# You could apply ffill separately if desired after the reindex that introduces NaNs.
new_index = ['one', 'two', 'three', 'four', 'five', 'six']
new_columns = ['color','price','new','object']
reindexed_frame = frame.reindex(new_index).reindex(columns=new_columns)
print("\nReindexed DataFrame 'frame' with new rows and column:\n", reindexed_frame)
```

### Dropping Elements from Series and DataFrames

To remove rows or columns from your data, pandas provides the `drop()` method. This method is handy because you can specify items to remove using their labels, making your code very readable. Importantly, `drop()` returns a *new* object with the specified items removed; it does not modify the original object in place by default.

*   **Dropping Rows**: Pass the label(s) of the row(s) you want to remove.
*   **Dropping Columns**: Pass the label(s) of the column(s) you want to remove, and crucially, set the `axis` parameter to `1` (or `'columns'`). By default, `drop()` operates on `axis=0` (rows).

```python
import pandas as pd
import numpy as np

# Example Series for dropping
ser = pd.Series([0.0, 1.0, 2.0, 3.0],
                index=['red','blue','yellow','white'])
print("Original Series 'ser':\n", ser)

# Dropping a single item from Series
print("\nSeries after dropping 'yellow':\n", ser.drop('yellow'))

# Dropping multiple items from Series
print("\nSeries after dropping 'blue' and 'white':\n", ser.drop(['blue','white']))

# Example DataFrame for dropping
frame = pd.DataFrame(np.arange(16).reshape((4,4)),
                     index=['red','blue','yellow','white'],
                     columns=['ball','pen','pencil','paper'])
print("\nOriginal DataFrame 'frame':\n", frame)

# Dropping single row from DataFrame
print("\nDataFrame after dropping row 'yellow':\n", frame.drop('yellow'))

# Dropping multiple rows from DataFrame
print("\nDataFrame after dropping rows 'blue' and 'yellow':\n", frame.drop(['blue','yellow']))

# Dropping multiple columns from DataFrame (axis=1)
print("\nDataFrame after dropping columns 'pen' and 'pencil':\n", frame.drop(['pen','pencil'], axis=1))
```

```python
import numpy as np
import pandas as pd

# ---------------------------------------------------------
# 1. SETUP: Creating Two DataFrames with Overlapping Labels
# ---------------------------------------------------------
# df_a has 'red' and 'pencil'; df_b has 'green' and 'mug'.
# These mismatches will demonstrate how Pandas handles missing data.

df_a = pd.DataFrame(np.arange(16).reshape((4,4)),
                    index=['red', 'blue', 'yellow', 'white'],
                    columns=['ball', 'pen', 'pencil', 'paper'])

df_b = pd.DataFrame(np.arange(12).reshape((4,3)),
                    index=['blue', 'green', 'white', 'yellow'],
                    columns=['mug', 'pen', 'ball'])

print("--- DataFrame A ---")
print(df_a)
print("\n--- DataFrame B ---")
print(df_b)

# ---------------------------------------------------------
# 2. DATA ALIGNMENT: Arithmetic between DataFrames
# ---------------------------------------------------------
# CONCEPT: Pandas aligns by BOTH index and column.
# Mismatched labels result in NaN (Not a Number).

print("\n--- Addition (df_a + df_b) ---")
# Result: 'red', 'green', 'pencil', 'paper', and 'mug' will be NaN
print(df_a + df_b)

print("\n--- Addition with Fill Value (df_a.add(df_b, fill_value=0)) ---")
# TEACHING TIP: This treats missing labels as 0 instead of NaN
print(df_a.add(df_b, fill_value=0))

# ---------------------------------------------------------
# 3. BROADCASTING: DataFrame vs. Series
# ---------------------------------------------------------
# CONCEPT: Operations between a DataFrame and a Series match
# the Series index to the DataFrame columns.

ser = pd.Series([0, 1, 2, 3], index=['ball', 'pen', 'pencil', 'paper'])
print("\n--- Series for Broadcasting ---")
print(ser)

print("\n--- Subtraction (df_a - ser) ---")
# The Series is subtracted from every row in the DataFrame
print(df_a - ser)

# ---------------------------------------------------------
# 4. ELEMENT-WISE OPERATIONS (Vectorization)
# ---------------------------------------------------------
# CONCEPT: Functions applied to the whole frame at once.

print("\n--- Square Root of df_a ---")
print(np.sqrt(df_a))

# ---------------------------------------------------------
# 5. MAPPING FUNCTIONS: Using Lambda and .apply()
# ---------------------------------------------------------
# CONCEPT: Applying a custom calculation to each column (axis=0)

# f calculates the range (Max - Min) for each column
f = lambda x: x.max() - x.min()

print("\n--- Column Range (Max - Min) ---")
print(df_a.apply(f))


# ---------------------------------------------------------
# 6. STATISTICS: Aggregation Functions
# ---------------------------------------------------------
# CONCEPT: By default, these operations happen "Column-wise" (axis=0).

print("\n--- Column Sums ---")
# Adds up every number in each column
print(df_a.sum())

print("\n--- Column Means (Averages) ---")
# Calculates the average for each column
print(df_a.mean())

# TEACHING TIP: Mention that they can change the 'axis'
# df_a.sum(axis=1) would sum the ROWS (red, blue, etc.)

# ---------------------------------------------------------
# 7. DATA INSPECTION: The describe() Method
# ---------------------------------------------------------
# CONCEPT: The "Quick Look." It provides 8 essential statistics at once.

print("\n--- Summary Statistics (describe) ---")
print(df_a.describe())
```

Supported Formats

The API supports a wide range of media, from simple text to complex databases:

- Binary & Office Formats: Microsoft Excel (read_excel), HDF5, Pickle.

```python
import pandas as pd

# Define the raw data
data = """1,5,2,3,cat
2,7,8,5,dog
3,3,6,7,horse
2,2,8,3,duck
4,4,2,1,mouse"""

# Write to a file named demo.csv
with open('demo.csv', 'w') as f:
    f.write(data)

print("File 'demo.csv' has been created successfully!")
```

Without with: You would have to manually type f.close() at the end. If you forget, the "pipe" stays open, which can lead to data corruption or memory leaks.

With with: This is called a Context Manager. As soon as the code inside the block finishes (or if there is an error), Python automatically "closes the pipe" for you.

Notice that the first animal (cat) and its numbers were promoted to the column titles. This is incorrect data parsing!

Reading in Chunks

Chunking is used when a file is too large to fit into the computer's memory (RAM). Instead of a DataFrame, read_csv with a chunksize returns an iterator.

```python
# Initialize an empty series to hold our results
out = pd.Series(dtype='int64')
i = 0

# Set chunksize=3 to process the file 3 rows at a time
pieces = pd.read_csv('demo.csv', chunksize=3, names=['white','red','blue','green','animal'])

for piece in pieces:
    # piece is a DataFrame containing only 3 rows
    out.at[i] = piece['white'].sum()
    i += 1

print("Sum of 'white' column per chunk:")
print(out)
```

# Reading and Writing HTML Files

pandas introduces tools for the Web. This is especially important for Web Scraping, as much of the world's data is embedded in HTML tables on websites.

Writing Data in HTML


The to_html() function automatically converts the internal structure of a DataFrame into nested HTML tags (<TH>, <TR>, <TD>). You don’t need to know HTML code to use this.

```python
import pandas as pd
import numpy as np

# Create a DataFrame
frame = pd.DataFrame(np.arange(4).reshape(2,2))

# This prints the raw HTML table code
print(frame.to_html())
```

To create a full web page, we wrap the table in standard HTML tags:

```python
# 1. Create a more complex DataFrame
frame_web = pd.DataFrame(np.random.random((4,4)),
                         index = ['white','black','red','blue'],
                         columns = ['up','down','right','left'])

# 2. Build the HTML structure
s = ['<HTML>', '<HEAD><TITLE>My DataFrame</TITLE></HEAD>', '<BODY>']
s.append(frame_web.to_html())
s.append('</BODY></HTML>')
html_string = ''.join(s)

# 3. Save to a file
with open('myFrame.html', 'w') as f:
    f.write(html_string)
```

Reading Data from an HTML File

The read_html() function performs the inverse: it parses an HTML page looking specifically for <table> tags.

read_html() always returns a list of DataFrames, because a single web page might contain multiple tables.

```python
# Reading from the file we just created
web_frames = pd.read_html('myFrame.html')

# Access the first table in the list using index [0]
print(web_frames[0])
```

Web Scraping from a URL

The same function can be used to scrape data directly from the Internet. By passing a URL, pandas downloads the page and extracts all tables.

```python
import pandas as pd
import requests
from io import StringIO

url = 'https://en.wikipedia.org/wiki/List_of_countries_and_dependencies_by_population'

# 1. Define a 'Header' to mimic a web browser
headers = {
    'User-Agent': 'Mozilla/5.0 (Windows NT 10.0; Win64; x64) AppleWebKit/537.36 (KHTML, like Gecko) Chrome/91.0.4472.124 Safari/537.36'
}

# 2. Get the page content using requests
response = requests.get(url, headers=headers)

# 3. Pass the HTML text to pandas
# We use StringIO to treat the string like a file
tables = pd.read_html(StringIO(response.text))

# 4. Display the result
print(f"Success! Found {len(tables)} tables.")
print(tables[0].head())
```

to_html(): Used for web developers to display DataFrames on websites.

read_html(): Used for data analysts to "scrape" data from the web.

Return Type: Remember that read_html returns a list; you must use [0] to get the first table.

Requirement: In a real environment, you often need the html5lib or lxml modules installed to make this work.

Reading Excel Files

Excel files often have multiple sheets. The read_excel() function allows you to target exactly which data you want to import.

If you don't specify a sheet, pandas will always read the first sheet in the workbook.

```python
# Reads the first sheet by default
df_default = pd.read_excel('data2.xlsx')
print(df_default)
```

```python
import pandas as pd

# 1. Create two different sets of data
data_sales = {'Item': ['Pens', 'Paper'], 'Amount': [100, 50]}
data_stock = {'Item': ['Pens', 'Paper'], 'In_Stock': [10, 25]}

df_sales = pd.DataFrame(data_sales)
df_stock = pd.DataFrame(data_stock)

# 2. Use ExcelWriter to save both to "Company_Data.xlsx"
with pd.ExcelWriter('Company_Data.xlsx') as writer:
    df_sales.to_excel(writer, sheet_name='Sales_Dept', index=False)
    df_stock.to_excel(writer, sheet_name='Warehouse', index=False)

print("Excel file 'Company_Data.xlsx' created with two sheets!")
```

Before reading data, students often don't know the names of the sheets. Use pd.ExcelFile to inspect the file without loading all the data into memory.

```python
# Create a pointer to the file
excel_file = pd.ExcelFile('Company_Data.xlsx')

# Check what sheets are inside
print(excel_file.sheet_names)
# Result: ['Sales_Dept', 'Warehouse']
```

Common Errors and How to Fix Them
Here are the "Triple-Check" rules for your students:

The "File Not Found" Error: Ensure the file name ends in .xlsx or .xls.

The "Sheet Not Found" Error: Ensure the sheet_name matches exactly (it is case-sensitive: "sheet1" is not the same as "Sheet1").

The "Module" Error: If Python says ImportError: Missing optional dependency 'openpyxl', you must install the engine that handles Excel files:

pip install openpyxl (for .xlsx)

pip install xlrd (for older .xls)
