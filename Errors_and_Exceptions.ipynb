1. ERRORS (Syntax Errors)

These happen when you write code that Python doesn't understand. The code won't even run.

```python
# Example 1: Missing colon
if 5 > 3
    print("Five is greater")
# SyntaxError: invalid syntax (missing ':')
```

```python
# Example 2: Wrong indentation
def greet():
print("Hello")
# IndentationError: expected an indented block
```

```python
# Example 3: Incomplete statement
result = 10 +
# SyntaxError: invalid syntax
```

```python
# Example 4: Missing closing parenthesis
print("Hello World"
# SyntaxError: '(' was never closed
```

```python
# Example 5: Too many closing parentheses
value = ((10 + 5) * 2))
# SyntaxError: unmatched ')'
```

```python
# Example 6: Mismatched parentheses in function
def add(a, b:
    return a + b
# SyntaxError: invalid syntax (missing ')')
```

```python
# Example 7: Wrong bracket types
my_list = [1, 2, 3)
# SyntaxError: closing parenthesis ')' does not match opening parenthesis '['
```

```python
# Example 8: Misspelled 'def'
deff greet():
    print("Hello")
# SyntaxError: invalid syntax
```

```python
# Example 9: Variable name misspelled
message = "Hello"
print(mesage)  # NameError: name 'mesage' is not defined
```

```python
# Example 10: Function name misspelled
def calculate_sum(a, b):
    return a + b

result = calcuate_sum(5, 3)  # NameError: name 'calcuate_sum' is not defined
```

2. Exceptions

Even if a statement or expression is syntactically correct, it may cause an error when an attempt is made to execute it.

Errors detected during execution are called exceptions and are not unconditionally fatal: you will soon learn how to handle them in Python programs

```python
# Example 1: Division by zero
print("Program started")
x = 10 / (1-1)  # ZeroDivisionError
print("This won't print")
```

```python
# Example 2: File not found
file = open("nonexistent.txt")  # FileNotFoundError
```

```python
# Example 4: Index out of range
fruits = ["apple", "banana"]
print(fruits[5])  # IndexError: list index out of range
```

3. Handling Exceptions

You cannot handle errors (must fix them first), but you can handle exceptions using try-except

If the programmer can determine the type of errors that can occur, then exception is used to do efficient programming and for solving the problem.

Some of the common exceptions that are defined in Python include:
- ImportError: Raised when an import statement fails
- IndexError: Raised when a sequence index is out of range
- NameError: Raised when a local or global name is not found
- IOError: Raised when an input/output operation fails
- ArithmeticError: Raised for numerical calculations
- OverflowError: Raised when a calculation exceeds max limit
- SyntaxError: Raised when the compiler encounters a syntax error
- ZeroDivisionError: Raised when there is division by zero
- ValueError: Raised when there is a mismatch in arguments.

```python
# Wrong spelling
import maths  # ImportError (correct is 'math')

# Correct way
import math
print(math.sqrt(16))  # Output: 4.0
```

```python
# List with only 3 items
fruits = ["apple", "banana", "orange"]

print(fruits[0])  # Works: apple
print(fruits[5])  # IndexError: list index out of range
```

```python
# Using a variable that doesn't exist
print(my_name)  # NameError: name 'my_name' is not defined

# Correct way
my_name = "Alice"
print(my_name)  # ✅ Works: Alice
```

```python
# Trying to open a file that doesn't exist
file = open("missing_file.txt")

# FileNotFoundError (a type of IOError)
```

```python
# Overflow
x = 10 ** 10000
print(x)
# ArithmeticError is more common in Python 2
```

```python
# Simple addition that should equal 0.3
result = 0.1 + 0.2
print(result)  # Output: 0.30000000000000004 (NOT 0.3!)

# Check if equal
print(0.1 + 0.2 == 0.3)  # Output: False
```

```python
# Adding 0.1 ten times
total = 0.0
for i in range(10):
    total += 0.1

print(total)  # Output: 0.9999999999999999 (NOT 1.0!)
print(total == 1.0)  # Output: False
```
