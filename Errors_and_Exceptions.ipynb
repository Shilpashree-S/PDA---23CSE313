3. Handling Exceptions

You cannot handle errors (must fix them first), but you can handle exceptions using try-except

If the programmer can determine the type of errors that can occur, then exception is used to do efficient programming and for solving the problem.

Some of the common exceptions that are defined in Python include:
- ImportError: Raised when an import statement fails
- IndexError: Raised when a sequence index is out of range
- NameError: Raised when a local or global name is not found
- IOError: Raised when an input/output operation fails
- ArithmeticError: Raised for numerical calculations
- OverflowError: Raised when a calculation exceeds max limit
- SyntaxError: Raised when the compiler encounters a syntax error
- ZeroDivisionError: Raised when there is division by zero
- ValueError: Raised when there is a mismatch in arguments.

```python
# Wrong spelling
import maths  # ImportError (correct is 'math')

# Correct way
import math
print(math.sqrt(16))  # Output: 4.0
```

Exception Handling is a way to deal with errors gracefully without crashing your program.

```python
# Program crashes!
print("Welcome to Calculator")
num1 = int(input("Enter number: "))  # User types "hello"
num2 = int(input("Enter number: "))
result = num1 + num2
print(f"Result: {result}")
print("Thank you!")  # This NEVER prints!
```

```python
# Program handles the error gracefully!
print("Welcome to Calculator")

try:
    num1 = int(input("Enter number: "))
    num2 = int(input("Enter number: "))
    result = num1 + num2
    print(f"Result: {result}")
except ValueError:
    print("Error: Please enter valid numbers!")

print("Thank you!")  # This ALWAYS prints!
```

Basic Syntax

```python
try:
    # Code that might cause an error
    risky_code()
except ExceptionType:
    # What to do if error occurs
    handle_error()
```

```python
# Simple exception handling
try:
    x = 10
    y = 0
    result = x / y  # This will cause ZeroDivisionError
    print(f"Result: {result}")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")

print("Program continues...")
```

```python
# Handling user input errors
try:
    age = int(input("Enter your age: "))
    print(f"You are {age} years old")
except ValueError:
    print("That's not a valid number!")

print("End of program")
```

```python
# Handling list index errors
fruits = ["apple", "banana", "orange"]

try:
    index = int(input("Enter index (0-2): "))
    print(f"Fruit: {fruits[index]}")
except IndexError:
    print("Error: Index out of range!")
except ValueError:
    print("Error: Please enter a number!")

print("Done!")
```

Multiple Except Blocks

```python
try:
    # risky code
except ExceptionType1:
    # handle error 1
except ExceptionType2:
    # handle error 2
except ExceptionType3:
    # handle error 3
```

```python
print("=== Simple Calculator ===")

try:
    num1 = float(input("Enter first number: "))
    num2 = float(input("Enter second number: "))
    operation = input("Enter operation (+, -, *, /): ")

    if operation == "+":
        result = num1 + num2
    elif operation == "-":
        result = num1 - num2
    elif operation == "*":
        result = num1 * num2
    elif operation == "/":
        result = num1 / num2
    else:
        print("Invalid operation!")
        result = None

    if result is not None:
        print(f"Result: {result}")

except ValueError:
    print("Error: Please enter valid numbers!")
except ZeroDivisionError:
    print("Error: Cannot divide by zero!")
except Exception as e:
    print(f"Unexpected error: {e}")

print("Calculator closed.")
```

Handle multiple exceptions the same way

```python
try:
    data = [10, 20, 30]
    index = int(input("Enter index: "))
    result = 100 / data[index]
    print(f"Result: {result}")

except (IndexError, ZeroDivisionError, ValueError) as e:
    print(f"Error occurred: {e}")

print("Done!")
```

THE else CLAUSE

The else block runs only if NO exception occurs.

Syntax

try:
    # code that might fail
except ExceptionType:
    # runs if error occurs
else:
    # runs if NO error occurs

THE finally CLAUSE

The finally block ALWAYS runs - whether error occurs or not!
