Ask the user to enter their marks (0–100).
Print:

"Distinction" if marks ≥ 75

"First class" if marks ≥ 60

"Pass" if marks ≥ 35

"Fail" otherwise

```python
marks = int(input("Enter the marks(1 - 100): "))
if marks >= 75:
  print("Distinction")
elif marks >= 60:
  print("First Class")
elif marks >= 35:
  print("Pass")
else:
  print("Fail")
```

Accept an integer and determine whether it is:

"Multiple of 3"

"Multiple of 5"

"Multiple of both"

"Not divisible by 3 or 5"

```python
num = int(input("Enter a number"))
if num % 3 == 0 and num % 5 == 0:
  print(f"{num} is multiple of 3 and 5")
elif num % 3 == 0:
  print(f"{num} is multiple of 3")
elif num % 5 == 0:
  print(f"{num} is multiple of 5")
else:
  print(f"{num} is not divisible by 3 and 5")
```

Take 5 numbers from the user (using loop). Count and display:

how many are even

how many are odd

```python
list_of_numbers = []
for i in range(5):
  num = int(input(f"Enter {i+1}th number"))
  list_of_numbers.append(num)

even_numbers = [num for num in list_of_numbers if num % 2 == 0]
odd_numbers = [num for num in list_of_numbers if num % 2 != 0]
print("Total even numbers:",len(even_numbers))
print("Total odd numbers:",len(odd_numbers))
```
