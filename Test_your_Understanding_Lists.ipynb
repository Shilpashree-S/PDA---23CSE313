## Part 1: Aliasing and Shared References (The Assignment Trap)

This section demonstrates the difference between creating an alias and creating a true copy.


```
# Exercise 1: Simple Assignment Creates an Alias
A = [10, 20, 30]
B = A # B is an alias (a pointer) to the list object A
B[0] = 99 # Modifying B changes the original object
print(f"1. A after changing B: {A}")
```



**Question:** What is the final value of the list `A` after changing `B[0]`? Why did this happen?

```
# Exercise 2: Using Slicing for a Safe Copy
data = [5, 6, 7]
backup = data[:]  # Safe Copy using slicing (creates a new list object)
backup.pop()
print(f"2. data after changing backup: {data}")
```

## Part 2: Mutability in Function Arguments (Side Effects)

This section explores how passing a mutable list to a function can lead to unexpected side effects on the original list.



```
# Exercise 3: Function Modifies the List In-Place
def remove_last(l):
  l.pop() # list.pop() modifies the list directly (in-place)

scores = [80, 85, 90]
remove_last(scores)
print(f"3. scores after function call: {scores}")
```

## Part 3: Loop Iteration Dangers (Changing size while looping)

This section demonstrates the risk of modifying a list's size while iterating over it.



```
# Exercise 5: Unsafe Removal (skips elements)
numbers_unsafe = [1, 2, 3, 4, 5]
# Modifying the list (numbers_unsafe.remove(n)) changes the list size and indices
# while the loop is reading based on the original size/indices.
for n in numbers_unsafe:
  if n % 2 == 1:
    numbers_unsafe.remove(n)
print(f"5. Unsafe list removal result: {numbers_unsafe}")
```


**Question:** Which number(s) were accidentally skipped or missed by the `remove()` logic?

-----



```
# Exercise 6: Safe Removal (iterating over a copy)
numbers_safe = [1, 2, 3, 4, 5]
safe_list = numbers_safe[:] # Copy for iteration
for n in safe_list: # Iterate over the stable copy
  if n % 2 == 1:
    numbers_safe.remove(n) # Modify the original list
print(f"6. Safe list removal result: {numbers_safe}")
```


**Question:** By iterating over a copy (`safe_list`), what are the contents of `numbers_safe` now?

## Part 4: Nested Lists and Shallow Copies (The inner list trap)

This section shows that simple slicing (`[:]`) only performs a **shallow copy**, meaning nested lists are still shared references.



```
# Exercise 7: Shallow Copy Pitfall
matrix_A = [[1, 2], [3, 4]]
matrix_B = matrix_A[:] # Shallow copy: creates a new outer list, but copies references to the inner lists
matrix_B[0][0] = 999 # Modifying the inner list at index 0
print(f"7. matrix_A[0][0] after changing B: {matrix_A[0][0]}")
```


**Question:** If you change an element in the nested list of `matrix_B`, does `matrix_A` also change? Why?

```
# Exercise 8: Shared Inner List Reference
names = [['A'], ['B']]
alias = names
copy_1 = names[:]
copy_1[0].append('X') # Appending to the inner list that is shared by all three variables
print(f"8. Alias list after modifying copy_1: {alias}")
```


**Question:** Does the list referenced by `alias` contain `'X'`? Explain the difference between `alias = names` and `copy_1 = names[:]` in this context.

## Part 5: Repetition Operator and Initialization

This section demonstrates the pitfall of using the `*` operator to initialize lists of mutable objects (like nested lists).



```
# Exercise 9: Unsafe Initialization (Shared references for inner lists)
rows = 3
cols = 2
# The inner list [0, 0] is created ONCE, and its reference is repeated 3 times.
grid_unsafe = [ [0] * cols ] * rows
grid_unsafe[0][0] = 1 # Changing the first element of the single shared inner list
print(f"9. grid_unsafe[1][0] after changing [0][0]: {grid_unsafe[1][0]}")
```


**Question:** Why did changing the value at `grid_unsafe[0][0]` also change the value at `grid_unsafe[1][0]`?

-----



```
# Exercise 10: Safe Initialization (List Comprehension)
rows = 3
cols = 2
# The list comprehension runs 3 times, creating a new inner list [0, 0] each time.
grid_safe = [[0] * cols for _ in range(rows)]
grid_safe[0][0] = 5
print(f"10. grid_safe[1][0] after changing [0][0]: {grid_safe[1][0]}")
```


**Question:** What is the advantage of using a **list comprehension** to initialize a 2D grid like this? What is `grid_safe[1][0]`?
