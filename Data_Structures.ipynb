## Mutability vs. Immutability: The Core Difference:

Before we dive into Lists, Tuples, and Dictionaries, we must understand the most critical concept that separates them: Mutability.

This determines whether an object can be changed after it has been created.


1. Mutable Objects:

    - Mutable objects can be modified, added to, or deleted from in place (without creating a new object).
    - Examples: Lists, Dictionaries, Sets

2. Immutable Objects:

    - Immutable objects cannot be changed after creation. If you try to modify them, Python actually creates an entirely new object in memory.
    - Examples: Tuples, Strings, Integers, Floats





### Lists

Lists are mutable, ordered sequences, commonly used to store collections of items that can be either homogeneous or heterogeneous.

Lists may be constructed in several ways:

  - Using a pair of square brackets to denote the empty list: []
  - Using square brackets, separating items with commas: [a], [a, b, c]
  - Using a list comprehension: [x for x in iterable]
  - Using the type constructor: list() or list(iterable)

```python
# 1. Using a pair of square brackets to denote the empty list:()
empty_shopping_list = []
print(f"1. Empty list constructed: {empty_shopping_list}")
print(f"Type check: {type(empty_shopping_list)}")
```

```python
# 2. Using square brackets, separating items with commas: [a], [a, b, c]
# Syntax: [item1, item2, item3, ...]
mixed_data_list = [
    "Python",         # String
    3.14,             # Float
    True,             # Boolean
    [1, 2, 3]         # Nested List
]

print(f"2. List with initial items: {mixed_data_list}")
print(f"Length: {len(mixed_data_list)}")
```

```python
# 3. Using a List Comprehension: `[x for x in iterable]`

# List comprehensions provide a concise, readable, and highly efficient way to
# create a list based on an existing iterable (like a range or another list),
#  often incorporating conditions or transformations.

# Scenario: Generate the square of every odd number from 1 to 10.

# Syntax: [EXPRESSION for ITEM in ITERABLE if CONDITION]
squares_of_odd_numbers = [
    x ** 2
    for x in range(1, 11)
    if x % 2 != 0
]

print(f"3. List created via list comprehension: {squares_of_odd_numbers}")
```

```python
## 4. Using the Type Constructor: `list()` or `list(iterable)`

# 4a. Empty list using constructor (Equivalent to [])
empty_list_b = list()
print(f"4a. Empty list via list(): {empty_list_b}")
print()
# The list() function takes any iterable (like a tuple, string, or range) and
# iterates over its elements one by one,
# creating a brand new list from those elements.

# Conversion from immutable tuple to mutable list:
# 4b. Converting a string to a list of characters
name_tuple = ("J", "a", "m", "e", "s")
list_from_tuple = list(name_tuple)
print(f"4b. List from tuple {name_tuple}: {list_from_tuple}")
print(type(list_from_tuple))
print()

# 4c. Converting a string to a list of characters
word = "COLAB"
list_from_string = list(word)
print("4c. String: ",word,"Datatype:",type(word))
print("List from string:",list_from_string,"Datatype:",type(list_from_string))
```

## **Accessing Elements by Index**

Indices are zero-based, meaning the first element is at index 0.

**Positive Indexing**: Access elements starting from the beginning of the list.

Syntax:
    
    list[i]

Note: The index i must be less than the length of the list.

**Negative Indexing**: Access elements starting from the end of the list.

Syntax:
    
    list[-i]

  - list[-1] always refers to the last element, list[-2] to the second-to-last, and so on.

```python
shopping_items = ["milk", "bread", "eggs", "coffee", "cheese"]
print(f"Original List: {shopping_items}")

# 1. Accessing the first item (Index 0)
first_item = shopping_items[0]
print(f"Index 0 (First Item): {first_item}")

# 2. Accessing the last item (Index -1)
last_item = shopping_items[-1]
print(f"Index -1 (Last Item): {last_item}")

# 3. Accessing the second-to-last item (Index -2)
second_to_last = shopping_items[-2]
print(f"Index -2: {second_to_last}")
```

**Modifying Elements by Index (Mutation)**:

Since lists are mutable, you can change the value of any element by assigning a new value to its index.



```python
# Our list: ["milk", "bread", "eggs", "coffee", "cheese"]

# Change the item at index 2 (which is "eggs")
shopping_items[2] = "bacon"
print(f"After modifying index 2: {shopping_items}")

# Change the item at the last position (Index -1, which is "cheese")
shopping_items[-1] = "yogurt"
print(f"After modifying index -1: {shopping_items}")
```

**Slicing Lists: Accessing Subsequences**:

Slicing allows you to extract a sequence of elements (a sub-list) from the original list. Slicing creates a new list (a shallow copy), so modifying the slice does not affect the original list.

Syntax:
    
    list[start:stop:step]

  - start: The index where the slice begins (inclusive). Defaults to 0.
  - stop: The index where the slice ends (exclusive—the item at this index is NOT included). Defaults to len(list).
  - step: The interval between items (e.g., 2 for every other item). Defaults to 1.




```python
data_records = [10, 20, 30, 40, 50, 60, 70, 80, 90]
print(f"\nOriginal Records: {data_records}")

# A. Basic Slice: Get elements from index 2 (30) up to (but not including) index 5 (60)
slice_a = data_records[2:5]
print(f"A. data_records[2:5] (Middle three): {slice_a}")

# B. Slice from the beginning (omitting 'start')
slice_b = data_records[:4]
print(f"B. data_records[:4] (First four): {slice_b}")

# C. Slice to the end (omitting 'stop')
slice_c = data_records[6:]
print(f"C. data_records[6:] (Last three): {slice_c}")

# D. Using a Step: Get every third item starting from the beginning
slice_d = data_records[::3]
print(f"D. data_records[::3] (Every third item): {slice_d}")

# E. Reversing the List (Advanced use of step)
reversed_list = data_records[::-1]
print(f"E. data_records[::-1] (Reversed): {reversed_list}")
```

## List Iteration: The for item in list Construct

The for loop is the standard and easiest way to access every element within a list or any other iterable collection in Python.

The construct **for item in list**: temporarily assigns each element of the list to the loop variable (item) sequentially, from first to last

```python
# Our sample list
sales_figures = [1050, 2200, 850, 3100, 1500]

# Calculate the total sales
total_sales = 0

print("--- Processing Sales Data ---")
for sale in sales_figures:
    # 'sale' is a copy of the list element's value
    total_sales += sale
    print(f"Processed sale: ${sale}. Current total: ${total_sales}")

print("\n--- Summary ---")
print(f"Final Total Sales: ${total_sales}")
```

**Important Warning**:

**Do Not Modify During Iteration**: You should NEVER add or remove items from a list while iterating over it using the simple for item in list syntax.

```python
# Original list with consecutive items to remove
tasks = ["buy milk", "buy milk", "buy milk", "clean floor", "walk dog"]
print(f"Original tasks: {tasks}")

# BAD PRACTICE: Attempting to remove items while iterating
for task in tasks:
    if task == "buy milk":
        tasks.remove(task)

print(f"\nResulting tasks (Incorrect Cleanup): {tasks}")
# EXPECTED FAILURE: The output is ['buy milk', 'clean floor', 'walk dog'],
# showing that two 'buy milk' items were skipped!


### The Correct (Pythonic) Solution

#If modification is required, the safest and cleanest approach is to create a new list or use a list comprehension.


# Correct way to filter items: create a new list
tasks = ["buy milk", "buy milk", "buy milk", "clean floor", "walk dog"]

filtered_tasks = [
    task
    for task in tasks
    if task != "buy milk"
]

print(f"Correctly Filtered Tasks: {filtered_tasks}")
# Result: ['clean floor', 'walk dog'] - All instances of 'buy milk' are gone.
```

## List Membership Testing: The in and not in Operators
The in construct is a boolean operator used to quickly check if a specific element exists within a collection (like a list, tuple, or dictionary). It returns True if the item is found and False otherwise.


Using the in Operator:

This is the preferred Pythonic way to test for existence in a list, offering much cleaner code than writing a loop to check every element manually.

```python
shopping_list = ["apples", "milk", "bread", "cheese", "coffee"]
item_to_find = "cheese"

# Test 1: Check if an item is present
if item_to_find in shopping_list:
    print(f"1. Success! '{item_to_find}' is on the list.")
else:
    print(f"1. '{item_to_find}' is not needed.")

# Test 2: Checking a different item
item_to_find_b = "bananas"
print(f"2. Is '{item_to_find_b}' on the list? {item_to_find_b in shopping_list}")
```

Using the not in Operator:

The not in operator is simply the negation of in. It is often used to ensure uniqueness before adding a new item to a list or to handle a condition where an item should not be present.

```python


## Practical Use Case: Data Validation


valid_colors = ["red", "green", "blue", "yellow"]

def process_color_input(color):
    """Checks if the input color is one of the valid options."""
    if color.lower() in valid_colors:
        print(f"'{color}' is a valid color. Processing...")
    else:
        print(f"'{color}' is NOT a valid color. Please try again.")

process_color_input("RED")
process_color_input("purple")


```

## Abnormal Termination of Loops:

**1. The break Statement**

The break statement is a control flow mechanism used to terminate the current loop prematurely.

  - The break statement may appear only syntactically nested in a for or while loop.
  - It terminates the nearest enclosing loop (the one it is directly inside).
  - If the loop is terminated by a break statement, the loop's else clause (if any) is not executed.
  - Execution continues at the statement immediately following the terminated loop.

Simply Put: **break** is an early exit. It halts the loop completely, regardless of whether the iteration sequence has finished.

Use Case:

Imagine searching a massive list of contacts (like a digital phonebook) for one specific person's name. The moment you find that name, you use break to immediately stop checking the remaining contacts, because any further searching is unnecessary.

**2. The continue Statement**

The continue statement is a control flow mechanism used to skip the rest of the current iteration and move to the next item in the loop sequence.


  - The continue statement may appear only syntactically nested in a for or while loop.
  - It continues with the next cycle of the nearest enclosing loop.
  - In a for loop, it advances the loop to the next item in the iterable.
  - In a while loop, it jumps back to the loop's conditional test.

Simply Put: **continue** is a skip-to-the-next-cycle command. It ignores the remaining code within the current loop body but keeps the loop running.

Use Case:

You are trying to process a batch of 100 student test scores, but 5 of those records are marked "Corrupted" or "Missing." You use continue to skip those 5 bad records and move on to the next valid score, ensuring your final average calculation is not ruined by bad data.

**3. The for...else Construct**

The else clause on a loop is a unique feature in Python that is tied directly to whether a loop completes naturally or is terminated by break.



  - The for loop may include an optional else clause.
  - The else clause is executed if and only if the loop terminates by exhausting the iterable (i.e., by iterating over every element).
  - The else clause is NOT executed if the loop is terminated by a break statement.

Simply Put: Use **for...else** when you need to confirm that a search operation failed. It provides a clean, single structure to answer the question: "Did the loop find what it was looking for, or did it check everything and fail?"

Use Case:

You are trying to check if a username is already taken in a database. You loop through all existing usernames. If you find a match, you break and say, "Name taken!" If the loop finishes exhaustively without hitting break, the else block executes, and you can definitively say: "The name is available because I checked every single existing user."

**4. The pass Statement**

The **pass** statement is a null operation—when Python executes it, nothing happens. It acts purely as a placeholder.

Use Cases:

- Incomplete Function/Class Definitions: You might define a function header (def my_function():) but haven't written the logic yet. Python requires an indented block, so pass fills that requirement.

- In Loops or Conditionals: You might be prototyping a loop where you only want to process certain items, but need to explicitly ignore all others.

Simply Put: **pass** allows you to write syntactically correct, empty blocks of code.





Functions for List: Modifying and Managing Data

Lists are powerful because they are mutable (changeable) and come with many built-in methods for modification and management.

**1. Methods for Adding Elements **

  - **list.append(x)**: Add an item to the end of the list.

  - **list.extend(iterable)**: Extend the list by appending all the items from the iterable.

  - **list.insert(i, x)**: Insert an item at a given position. The first argument is the index of the element before which to insert, so a.insert(0, x) inserts at the front of the list, and a.insert(len(a), x) is equivalent to a.append(x).


**2. Methods for Removing Elements**

  - **list.remove(x)**: Remove the first item from the list whose value is equal to x. It raises a ValueError if there is no such item.

  - **list.pop([i])**: Remove the item at the given position in the list, and return it. If no index is specified, a.pop() removes and returns the last item in the list. It raises an IndexError if the list is empty or the index is outside the list range.

  - **list.clear()**: Remove all items from the list.

**3. Utility, Search, and Counting**

  - **list.index(x, [start, end])**: Return zero-based index of the first occurrence of x in the list. Raises a ValueError if there is no such item.

    The optional arguments start and end are interpreted as in the slice notation and are used to limit the search to a particular subsequence of the list. The returned index is computed relative to the beginning of the full sequence rather than the start argument.

  - **list.count(x):**: Return the number of times x appears in the list.

**4. Reordering and Copying**

    - **list.sort()**: Sort the items of the list in place. Use keyword arguments key (for custom sorting logic) and reverse=True (for descending order).

    - **list.reverse()**: Reverse the elements of the list in place.

    - **list.copy()**: Return a shallow copy of the list.
