1. Sales Data Analysis

You have sales data for a store for 7 days: [250, 300, 150, 400, 350, 500, 450].

- Create a NumPy array from this data.

- Find total sales and average sales.

- Identify days where sales exceeded 350.

```python
# Create a Numpy array from this data.
sales_data = np.array([250, 300, 150, 400, 350, 500, 450])
print("Sales data: ",sales_data)

# Find total sale and average sales.
total_sales = np.sum(sales_data)
avg_sales = np.mean(sales_data)

print("Total sales: ",total_sales)
print("Average sales: ",avg_sales)

#Identify days where sales exceeded 350
print("Data of days where sales exceeded 350",sales_data[sales_data > 350])
```

3.Image Pixel Manipulation

You are given a 5×5 grayscale image represented as integers 0–255.

- Create a 5×5 array filled with 128.

- Increase brightness by 50 using a scalar operation.

- Ensure the values do not exceed 255.

```python
# Create a 5×5 array filled with 128.
image_pix_array = np.full((5,5),128)
print("Initial Image pixel array:\n",image_pix_array)

# Increase brightness by 50 using a scalar operation.
bright_arr = image_pix_array + 50
print("Result after scalar addition: \n",bright_arr)

# Ensure the values do not exceed 255.
clipped_array = np.clip(bright_arr, 0, 255)
print("Clipped array:\n", clipped_array)
```

Note:

The numpy.clip() function in Python is used to limit the values in a NumPy array to a specified range. It takes an array, a minimum value, and a maximum value as input. Any element in the array that is less than the minimum value will be replaced by the minimum value, and any element greater than the maximum value will be replaced by the maximum value.

Syntax:

numpy.clip(a, a_min, a_max, out=None)

20. Counting Positive, Negative, and Zero Values (1D)

Given an array of integers, iterate through it and count:

  - how many numbers are positive

  - how many are negative

  - how many are zero

  - Print all three counts.

```python
a = np.array([3, -1, 0, 7, -9, 12, 0])

pos = neg = zero = 0

for i in a:
    if i > 0:
        pos += 1
    elif i < 0:
        neg += 1
    else:
        zero += 1

print("Positive:", pos)
print("Negative:", neg)
print("Zero:", zero)
```

24. Flatten and Find Maximum Using Iteration

    for item in A.flat:

25. You are given the following 3×4 matrix:

            A = np.array([
                  [10, 20, 30, 40],
                  [5, 15, 45, 25],
                  [8, 18, 28, 38]
               ])


- Define a function that takes a 1D array (a row) and returns:

      max(row) – min(row)


- Use np.apply_along_axis() with axis=1 to compute the range of each row.

```python
import numpy as np

A = np.array([
    [10, 20, 30, 40],
    [5, 15, 45, 25],
    [8, 18, 28,38]
])

def row_range(x):
    return x.max() - x.min()

result = np.apply_along_axis(row_range, axis=1, arr=A)
print(result)
```

26. Column-wise Product Using apply_along_axis()

Consider the following 2×4 matrix:

        A = np.array([
              [2, 3, 4, 5],
              [6, 7, 8, 9]
            ])



- Write a custom function that returns the product of all elements in a 1D vector.

- Use np.apply_along_axis() with axis=0 to compute the product of each column.

```python
A = np.array([
    [2, 3, 4, 5],
    [6, 7, 8, 9]
])

def product(x):
    p = 1
    for i in x:
        p *= i
    return p

result = np.apply_along_axis(product, axis=0, arr=A)
print(result)
```

27. Normalize Each Row Using apply_along_axis()

You are given this 3×3 matrix of test scores:

Write the following function:

      def normalize(x):

28. Count Even Numbers in Each Column

You are given the following 3×4 matrix:

      A = np.array([
        [12, 15, 20, 33],
        [22, 35, 44, 55],
        [13, 66, 77, 88]
    ])



- Write a custom function that receives a 1D array (a column), counts how many values are even, and returns that count.

- Use:

      np.apply_along_axis(func, axis=0, arr=A)


to compute even counts for each column.

```python
A = np.array([
    [12, 15, 20, 33],
    [22, 35, 44, 55],
    [13, 66, 77, 88]
])

def count_even(x):
    count = 0
    for i in x:
        if i % 2 == 0:
            count += 1
    return count

result = np.apply_along_axis(count_even, axis=0, arr=A)
print(result)
```

30. Calculate the Median Deviation for Each Column

You are given the following 4×3 matrix representing daily measurements from three sensors:

      A = np.array([
          [12, 18, 25],
          [15, 20, 22],
          [14, 19, 28],
          [16, 21, 24]
      ])



- Define a custom function that takes a 1D array (a column) and returns:

      median deviation = median(column) - min(column)


- Use np.apply_along_axis() with axis=0 to compute the median deviation of each column.

- Print the resulting array.

```python
import numpy as np

A = np.array([
    [12, 18, 25],
    [15, 20, 22],
    [14, 19, 28],
    [16, 21, 24]
])
def median_deviation(x):
    return np.median(x) - np.min(x)
result = np.apply_along_axis(median_deviation, axis=0, arr=A)
print(result)
```
